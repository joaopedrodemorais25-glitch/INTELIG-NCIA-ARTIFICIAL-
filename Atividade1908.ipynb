7  Crie dois arrays de numeros aleatorios e encontre a correlacao entre eles.

```python
import numpy as np

array1 = np.random.rand(20)
array2 = np.random.rand(20)

media1 = np.mean(array1)
media2 = np.mean(array2)

covariancia = np.sum((array1 - media1) * (array2 - media2)) / len(array1)
desvio1 = np.std(array1)
desvio2 = np.std(array2)

correlacao = covariancia / (desvio1 * desvio2)
print("Coeficiente de correlação:", correlacao)
```

10. Calcule a integral dupla da funcao f(x, y) = xy + x
2 no intervalo [0, 2]
para x e [0, 1] para y.

```python
from scipy import integrate

def f(x, y):
    return x*y + x**2

def integral_em_y(x):
    def f_com_x_fixo(y):
        return f(x, y)
    resultado_y, erro_y = integrate.quad(f_com_x_fixo, 0, 1)
    return resultado_y

resultado, erro = integrate.quad(integral_em_y, 0, 2)
print("Resultado da integral dupla:", resultado)
```

11. Crie um grafico de barras empilhadas para tres categorias com tres grupos
de valores

```python
import matplotlib.pyplot as plt

categorias = ['A', 'B', 'C']
grupo1 = [3, 5, 2]
grupo2 = [2, 4, 3]
grupo3 = [1, 2, 4]

# Somar grupo1 com grupo2, posição por posição
soma_grupo1_grupo2 = []
for i in range(len(grupo1)):
    soma_grupo1_grupo2.append(grupo1[i] + grupo2[i])

plt.bar(categorias, grupo1, label='Grupo 1')
plt.bar(categorias, grupo2, bottom=grupo1, label='Grupo 2')
plt.bar(categorias, grupo3, bottom=soma_grupo1_grupo2, label='Grupo 3')
plt.legend()
plt.title("Barras Empilhadas")
plt.show()
```

12. Plote um grafico de linha com uma funcao quadratica e ajuste os limites
do eixo para focar nos valores x = −5 a x = 5

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(-10, 10, 200)
y = x ** 2

plt.plot(x, y)
plt.xlim([-5, 5])
plt.title("Função Quadrática (foco em x = -5 a 5)")
plt.show()
```

13. Gere um grafico de dispersao 3D para 100 pontos aleatorios utilizando
Matplotlib.

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.random.rand(100)
y = np.random.rand(100)
z = np.random.rand(100)

fig = plt.figure()
eixo = fig.add_subplot(projection='3d')
eixo.scatter(x, y, z)

plt.title("Dispersão 3D")
plt.show()
```

16. Crie um grafico de linha com tıtulos para cada eixo e adicione um rotulo
de texto no ponto x = 5

18. Crie um grafico de barras com valores positivos e negativos e use diferentes
cores para cada tipo de valor.

```python
import matplotlib.pyplot as plt

categorias = ['A', 'B', 'C', 'D']
valores = [5, -3, 7, -2]

cores = []
for valor in valores:
    if valor >= 0:
        cores.append('green')
    else:
        cores.append('red')

plt.bar(categorias, valores, color=cores)
plt.axhline(0, color='black')
plt.title("Valores Positivos e Negativos")
plt.show()
```

19. Plote um grafico de linha com duas series e adicione uma legenda para
diferencia-las.

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 10, 100)

plt.plot(x, np.sin(x), label='Seno')
plt.plot(x, np.cos(x), label='Cosseno')
plt.legend()
plt.title("Duas Séries")
plt.show()
```
