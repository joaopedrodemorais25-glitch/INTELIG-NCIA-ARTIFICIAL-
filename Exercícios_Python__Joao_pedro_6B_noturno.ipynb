questao 21 busca binária para encontrar um elemento em uma
lista ordenada

```python
lista = [1, 2, 3, 4, 5, 6, 7, 8, 9]
numero = int(input("Digite um número: "))
inicio = 0
fim = len(lista) - 1
while inicio <= fim:
    meio = (inicio + fim) // 2

    if lista[meio] == numero:
        print("Achou!")
        break
    elif lista[meio] < numero:
        inicio = meio + 1
    else:
        fim = meio - 1
else:
    print("Não achou!")
```

questao 30 cria um jogo da forca

```python
palavra = "cao "
tentativas = 6
acertos = ""

while tentativas > 0:
    letra = input("Digite uma letra: ")

    if letra in palavra:
        acertos = acertos + letra
        print("Acertou!")
    else:
        tentativas = tentativas - 1
        print("Errou! Tentativas:", tentativas)

    palavra2 = ""

    for letra in palavra:
        if letra in acertos:
            palavra2 = palavra2 + letra
        else:
            palavra2 = palavra2 + "_"

    print(palavra2)

    if palavra2 == palavra:
        print("Você ganhou!")
        break

if tentativas == 0:
    print("Você perdeu! A palavra era:", palavra)
```

questao 31 ordena por bolha

```python
lista = [5, 2, 8, 1, 3]

for i in range(len(lista)):
    for j in range(len(lista) - 1):
        if lista[j] > lista[j + 1]:
            temp = lista[j]
            lista[j] = lista[j + 1]
            lista[j + 1] = temp

print(lista)
```

questao 46 calcula a media e o desvio padrao de uma lista

```python
lista = [2, 4, 6, 8, 10]

media = sum(lista) / len(lista)

soma = 0

for numero in lista:
    soma = soma + (numero - media) ** 2

desvio = (soma / len(lista)) ** 0.5

print("Média:", media)
print("Desvio padrão:", desvio)
```

questao 48 ele pega um numero por vez e coloca por ordem

```python
lista = [5, 2, 8, 1, 3]

for i in range(1, len(lista)):
    numero = lista[i]
    j = i - 1

    while j >= 0 and lista[j] > numero:
        lista[j + 1] = lista[j]
        j = j - 1

    lista[j + 1] = numero

print(lista)
```
